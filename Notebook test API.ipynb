```python
import pandas as pd
import requests
```

```python
CITY_COORDS = {
    "Paris": (48.8566, 2.3522),
    "Lyon": (45.7640, 4.8357),
    "Marseille": (43.2965, 5.3698),
    "Bordeaux": (44.8378, -0.5792),
    "Lille": (50.6292, 3.0573),
    "Nantes": (47.2184, -1.5536),
    "Toulouse": (43.6047, 1.4442),
    "Nice": (43.7102, 7.2620)
}
```

```python
START_DATE = "2024-05-01"
END_DATE = "2024-08-31"
```

```python
# 3. Créer la fonction fetch_weather_for_city()
def fetch_weather_for_city(city, lat, lon):

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": START_DATE,
        "end_date": END_DATE,
        "daily": [
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum",
            "wind_speed_10m_max"
        ],
        "timezone": "Europe/Paris"
    }

```

```python
# Appel de l'API
response = requests.get(url, params=params, timeout=30)
```

```python
weather_data = []

# Parcourir toutes les villes
for city, (lat, lon) in CITY_COORDS.items():
    print(f"Récupération des données météo pour {city}...")
    # Récupérer les données météo de la ville
    city_weather = fetch_weather_for_city(city, lat, lon)
    # Ajouter le DataFrame à la liste
    weather_data.append(city_weather)

# Concaténer les données de toutes les villes
weather = pd.concat(weather_data, ignore_index=True)

# Vérification des données
print("\n--- Vérification des données météo ---")
print("Dimensions du DataFrame :", weather.shape)
print("\nVilles présentes :")
print(weather["city"].unique())
print("\nDate minimale :", weather["date"].min())
print("Date maximale :", weather["date"].max())
```
